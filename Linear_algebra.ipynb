## 3. Vector Calculations

Vectors are one-dimensional arrays of numbers that represent quantities such as direction, movement, or data points. In linear algebra, vectors are used to model many real-world relationships, from coordinates in space to features in a dataset. Python and NumPy make it easy to perform vector-based calculations efficiently.


### 3.1 Adding Vectors

Vector addition is done by adding corresponding elements from two vectors. This operation combines two vectors while preserving their direction and magnitude in a structured way.

```python
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

vector_sum = vector_a + vector_b

print("Vector A:", vector_a)
print("Vector B:", vector_b)
print("Vector Sum:", vector_sum)
```

**Interpretation:** The corresponding elements of the two vectors are added together. Therefore, the sum of `[1, 2, 3]` and `[4, 5, 6]` is `[5, 7, 9]`.

### 3.2 Multiplying a Vector by a Scalar


Scalar multiplication means multiplying each element of a vector by the same number. This changes the vector’s size without changing its direction, unless the scalar is negative, in which case the direction is reversed.


```python
vector = np.array([1, 2, 3])
scalar = 3

result = scalar * vector

print("Vector:", vector)
print("Scalar:", scalar)
print("Result:", result)
```

**Interpretation:** Each element of the vector was multiplied by the scalar value 3. Therefore, `[1, 2, 3]` becomes `[3, 6, 9]`.

### 3.3 Dot Product and Vector Magnitude

The dot product measures how closely two vectors align with each other by multiplying corresponding elements and summing the results. The norm of a vector represents its length or magnitude, which is useful in many mathematical and machine learning applications.

```python
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

dot_product = np.dot(vector_a, vector_b)
norm_a = np.linalg.norm(vector_a)

print("Vector A:", vector_a)
print("Vector B:", vector_b)
print("Dot Product:", dot_product)
print("Norm of Vector A:", norm_a)
```

**Interpretation:** The dot product of the two vectors is 32. The norm of Vector A is approximately 3.74, which represents its magnitude.

## 4. Matrix Operations

A matrix is a rectangular arrangement of numbers in rows and columns. Matrices are fundamental in linear algebra because they can represent systems of equations, datasets, transformations, and many other structures in data science and engineering.


### 4.1 Adding and Multiplying Matrices

Matrix addition is performed element by element, while matrix multiplication follows a different rule that combines rows and columns. Matrix multiplication is especially important because it allows one matrix to transform or interact with another in a meaningful way.


```python
matrix_a = np.array([
    [1, 2],
    [3, 4]
])

matrix_b = np.array([
    [5, 6],
    [7, 8]
])

matrix_sum = matrix_a + matrix_b
matrix_product = matrix_a @ matrix_b

print("Matrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)

print("\nMatrix Addition:")
print(matrix_sum)

print("\nMatrix Multiplication:")
print(matrix_product)
```

### 4.2 Transpose and Inverse of a Matrix

The transpose of a matrix swaps its rows and columns. The inverse of a square matrix is another matrix that, when multiplied with the original, produces the identity matrix. Inverse matrices are useful for solving systems of equations.


```python
matrix = np.array([
    [1, 2],
    [3, 4]
])

transpose = matrix.T
inverse = np.linalg.inv(matrix)

print("Original Matrix:")
print(matrix)

print("\nTranspose:")
print(transpose)

print("\nInverse:")
print(inverse)
```

### 4.3 Matrix Rank

The rank of a matrix indicates the number of linearly independent rows or columns it contains. This measure helps determine whether a matrix has full rank and whether a system of equations has a unique solution.

```python
matrix = np.array([
    [1, 2],
    [3, 4]
])

rank = np.linalg.matrix_rank(matrix)

print("Matrix:")
print(matrix)

print("\nRank:", rank)
```

## 5. Eigenvalues and Eigenvectors
Eigenvalues and eigenvectors are important concepts in linear algebra that describe how a matrix transforms space. For a given square matrix, an eigenvector is a non-zero vector whose direction remains unchanged after multiplication by the matrix, while the eigenvalue tells how much the vector is scaled.


### 5.1 Understanding the Concept

When a matrix acts on an eigenvector, the output is simply the same vector multiplied by a scalar value. This property makes eigenvalues and eigenvectors especially useful in areas such as physics, computer graphics, and machine learning.

### 5.2 Computing in Python

Using NumPy, we can efficiently compute eigenvalues and eigenvectors for a square matrix. This is useful when analyzing the behavior of linear transformations and understanding the structure of a matrix.


```python
matrix = np.array([
    [2, 0],
    [0, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(matrix)

print("Matrix:")
print(matrix)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

### 5.3 Checking and Interpreting the Results

The eigenvalues and eigenvectors returned by NumPy provide information about how the matrix stretches or compresses space. The eigenvalues represent the scale factors, and the eigenvectors show the directions that are preserved under the transformation.
